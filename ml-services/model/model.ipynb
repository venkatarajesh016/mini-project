````python
# ===============================
# 1. IMPORT LIBRARIES
# ===============================
import pandas as pd
import torch
import torch.nn.functional as F
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics.pairwise import cosine_similarity
from torch_geometric.nn import GCNConv

# fill missing values
df['singer'] = df['singer'].fillna("Unknown")
df['artist'] = df['artist'].fillna("Unknown")
df['album'] = df['album'].fillna("Unknown")
df['year'] = df['year'].fillna(-1)
df.rename(columns={"Genre": "genre"}, inplace=True)
df['genre'] = df['genre'].fillna("Unknown")
df = df.reset_index(drop=True)

print("Total songs:", len(df))

# ===============================
# 3. CREATE SONG-ID MAPPING
# ===============================
songs = df['title'].tolist()

song_to_id = {song:i for i,song in enumerate(songs)}
id_to_song = {i:song for song,i in song_to_id.items()}

# ===============================
# 4. BUILD GRAPH EDGES + REASONS
# ===============================
edges = []
edge_reason = {}
edge_weight = {}

for i in range(len(df)):
    for j in range(i+1, len(df)):

        weight = 0
        reasons = []

        # same singer (strong)
        if df.loc[i,'singer'] == df.loc[j,'singer']:
            weight += 4
            reasons.append("Singer: " + str(df.loc[i,'singer']))

        # same artist
        if df.loc[i,'artist'] == df.loc[j,'artist']:
            weight += 3
            reasons.append("Artist: " + str(df.loc[i,'artist']))

        # same genre
        if df.loc[i,'genre'] == df.loc[j,'genre']:
            weight += 2
            reasons.append("Genre: " + str(df.loc[i,'genre']))

        # same album (weakest)
        if df.loc[i,'album'] == df.loc[j,'album']:
            weight += 1
            reasons.append("Album: " + str(df.loc[i,'album']))

        if weight > 0:

            edges.append([i,j])
            edges.append([j,i])

            edge_reason[(i,j)] = ", ".join(reasons)
            edge_reason[(j,i)] = ", ".join(reasons)

            edge_weight[(i,j)] = weight
            edge_weight[(j,i)] = weight

print("Total edges:", len(edges))

# ===============================
# 5. CREATE NODE FEATURES
# ===============================
singer_encoder = LabelEncoder()
artist_encoder = LabelEncoder()
album_encoder = LabelEncoder()

df['singer_id'] = singer_encoder.fit_transform(df['singer'])
df['artist_id'] = artist_encoder.fit_transform(df['artist'])
df['album_id'] = album_encoder.fit_transform(df['album'])

features = df[['singer_id','artist_id','album_id','year']].values

x = torch.tensor(features, dtype=torch.float)

# ===============================
# 6. CREATE EDGE INDEX FOR GNN
# ===============================
edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()

# ===============================
# 7. DEFINE GNN MODEL
# ===============================
class MusicGNN(torch.nn.Module):

    def __init__(self, input_dim, hidden_dim=16, output_dim=8):
        super().__init__()

        self.conv1 = GCNConv(input_dim, hidden_dim)
        self.conv2 = GCNConv(hidden_dim, output_dim)

    def forward(self, x, edge_index):

        x = self.conv1(x, edge_index)
        x = F.relu(x)

        x = self.conv2(x, edge_index)

        return x

# ===============================
# 8. TRAIN GNN
# ===============================
model = MusicGNN(input_dim=x.shape[1])

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

print("\nTraining GNN...")

for epoch in range(100):

    model.train()

    optimizer.zero_grad()

    embeddings = model(x, edge_index)

    loss = torch.mean(embeddings**2)

    loss.backward()

    optimizer.step()

    if epoch % 20 == 0:
        print("Epoch:", epoch, "Loss:", loss.item())

# ===============================
# 9. GET FINAL EMBEDDINGS
# ===============================
model.eval()

embeddings = model(x, edge_index).detach().numpy()

print("\nGNN training complete")

# ===============================
# 10. RECOMMENDATION FUNCTION
# ===============================
def recommend(song_name):

    if song_name not in song_to_id:
        print("Song not found")
        return

    idx = song_to_id[song_name]

    similarity = cosine_similarity(
        embeddings[idx].reshape(1,-1),
        embeddings
    )[0]

    sorted_indices = similarity.argsort()[::-1]

    selected = []
    used = set()

    # 1️⃣ Same Singer
    for i in sorted_indices:
        if i != idx and df.loc[i,'singer'] == df.loc[idx,'singer']:
            selected.append((i, "Same Singer"))
            used.add(i)
            break

    # 2️⃣ Same Genre
    for i in sorted_indices:
        if i != idx and i not in used and df.loc[i,'genre'] == df.loc[idx,'genre']:
            selected.append((i, "Same Genre"))
            used.add(i)
            break

    # 3️⃣ Same Artist
    for i in sorted_indices:
        if i != idx and i not in used and df.loc[i,'artist'] == df.loc[idx,'artist']:
            selected.append((i, "Same Artist"))
            used.add(i)
            break

    # 4️⃣ Same Album
    for i in sorted_indices:
        if i != idx and i not in used and df.loc[i,'album'] == df.loc[idx,'album']:
            selected.append((i, "Same Album"))
            used.add(i)
            break

    # 5️⃣ Pure Similarity
    for i in sorted_indices:
        if i != idx and i not in used:
            selected.append((i, "Similar Embedding"))
            break
    print("\nRecommendations for:", song_name)
    print("="*50)

    for i, reason in selected:

        print("Recommended Song:", id_to_song[i])
        print("Singer:", df.loc[i, 'singer'])
        print("Artist:", df.loc[i, 'artist'])
        print("Genre:", df.loc[i, 'genre'])
        print("Album:", df.loc[i, 'album'])
        print("Reason:", reason)
        print("-"*50)
recommend("Chamka Chamka")
def evaluate_precision_at_k(k=5):

    total_precision = 0
    total_songs = 0

    for song in songs:

        idx = song_to_id[song]

        similarity = cosine_similarity(
            embeddings[idx].reshape(1,-1),
            embeddings
        )[0]

        sorted_indices = similarity.argsort()[::-1][1:k+1]

        relevant = 0

        for i in sorted_indices:

            if (
                df.loc[i,'singer'] == df.loc[idx,'singer'] or
                df.loc[i,'artist'] == df.loc[idx,'artist'] or
                df.loc[i,'genre'] == df.loc[idx,'genre'] or
                df.loc[i,'album'] == df.loc[idx,'album']
            ):
                relevant += 1

        precision = relevant / k
        total_precision += precision
        total_songs += 1

    avg_precision = total_precision / total_songs

    print("Precision@",k,"=", round(avg_precision,3))
````

````python
def full_evaluation(k=5):
    precisions = []
    recalls = []

    for song in songs:
        idx = song_to_id[song]

        # Count total relevant songs in dataset
        total_relevant = sum(
            1 for i in range(len(df))
            if i != idx and (
                df.loc[i,'singer'] == df.loc[idx,'singer'] or
                df.loc[i,'artist'] == df.loc[idx,'artist'] or
                df.loc[i,'genre']  == df.loc[idx,'genre']  or
                df.loc[i,'album']  == df.loc[idx,'album']
            )
        )

        similarity = cosine_similarity(
            embeddings[idx].reshape(1,-1), embeddings
        )[0]

        top_k = similarity.argsort()[::-1][1:k+1]

        hits = sum(
            1 for i in top_k
            if (
                df.loc[i,'singer'] == df.loc[idx,'singer'] or
                df.loc[i,'artist'] == df.loc[idx,'artist'] or
                df.loc[i,'genre']  == df.loc[idx,'genre']  or
                df.loc[i,'album']  == df.loc[idx,'album']
            )
        )

        precisions.append(hits / k)
        recalls.append(hits / total_relevant if total_relevant > 0 else 0)

    avg_precision = sum(precisions) / len(precisions)
    avg_recall    = sum(recalls)    / len(recalls)
    f1 = (2 * avg_precision * avg_recall / (avg_precision + avg_recall)
          if (avg_precision + avg_recall) > 0 else 0)

    print(f"Precision@{k} = {avg_precision:.3f}")
    print(f"Recall@{k}    = {avg_recall:.3f}")
    print(f"F1-Score@{k}  = {f1:.3f}")

full_evaluation(k=5)
full_evaluation(k=10)
````

````python
# ===============================
# ADVANCED GNN WITH IMPROVEMENTS
# ===============================
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
from torch_geometric.nn import SAGEConv
import numpy as np

# Data preparation
df['singer'] = df['singer'].fillna("Unknown")
df['artist'] = df['artist'].fillna("Unknown")
df['album']  = df['album'].fillna("Unknown")
df['year']   = df['year'].fillna(-1)
df.rename(columns={"Genre": "genre"}, inplace=True)
df['genre']  = df['genre'].fillna("Unknown")
df = df.reset_index(drop=True)

print("Total songs:", len(df))

# Song mapping
songs      = df['title'].tolist()
song_to_id = {song: i for i, song in enumerate(songs)}
id_to_song = {i: song for song, i in song_to_id.items()}

# Build edges with weights
edges = []
edge_weight = {}

for i in range(len(df)):
    for j in range(i + 1, len(df)):
        weight = 0
        if df.loc[i, 'singer'] == df.loc[j, 'singer'] and df.loc[i, 'singer'] != "Unknown":
            weight += 4
        if df.loc[i, 'artist'] == df.loc[j, 'artist'] and df.loc[i, 'artist'] != "Unknown":
            weight += 3
        if df.loc[i, 'genre'] == df.loc[j, 'genre'] and df.loc[i, 'genre'] != "Unknown":
            weight += 2
        if df.loc[i, 'album'] == df.loc[j, 'album'] and df.loc[i, 'album'] != "Unknown":
            weight += 1
        if weight > 0:
            edges.append([i, j])
            edges.append([j, i])
            edge_weight[(i, j)] = weight
            edge_weight[(j, i)] = weight

print("Total edges:", len(edges))

# Node features (normalized)
singer_encoder = LabelEncoder()
artist_encoder = LabelEncoder()
album_encoder  = LabelEncoder()
genre_encoder  = LabelEncoder()

df['singer_id'] = singer_encoder.fit_transform(df['singer'])
df['artist_id'] = artist_encoder.fit_transform(df['artist'])
df['album_id']  = album_encoder.fit_transform(df['album'])
df['genre_id']  = genre_encoder.fit_transform(df['genre'])

df['year_norm'] = df['year'].astype(float)
df['year_norm'] = df['year_norm'].replace(-1, df['year_norm'][df['year_norm'] != -1].mean())

features = df[['singer_id', 'artist_id', 'album_id', 'genre_id', 'year_norm']].values
scaler   = StandardScaler()
features = scaler.fit_transform(features)

x          = torch.tensor(features, dtype=torch.float)
edge_index = torch.tensor(edges, dtype=torch.long).t().contiguous()

print("Feature shape:", x.shape)

# Improved GNN Model
class MusicGNN(nn.Module):
    def __init__(self, input_dim, hidden_dim=128, output_dim=64):
        super().__init__()
        self.conv1 = SAGEConv(input_dim, hidden_dim)
        self.conv2 = SAGEConv(hidden_dim, hidden_dim)
        self.conv3 = SAGEConv(hidden_dim, output_dim)
        self.bn1   = nn.BatchNorm1d(hidden_dim)
        self.bn2   = nn.BatchNorm1d(hidden_dim)
        self.dropout = nn.Dropout(p=0.2)

    def forward(self, x, edge_index):
        x = self.conv1(x, edge_index)
        x = self.bn1(x)
        x = F.relu(x)
        x = self.dropout(x)
        x = self.conv2(x, edge_index)
        x = self.bn2(x)
        x = F.relu(x)
        x = self.dropout(x)
        x = self.conv3(x, edge_index)
        x = F.normalize(x, p=2, dim=1)
        return x

# Contrastive Loss
def contrastive_loss(embeddings, edge_index, num_neg=5, margin=1.0):
    src = edge_index[0]
    dst = edge_index[1]
    pos_dist = F.pairwise_distance(embeddings[src], embeddings[dst])
    pos_loss = pos_dist.pow(2).mean()
    neg_idx  = torch.randint(0, embeddings.size(0), (src.size(0) * num_neg,))
    src_rep  = src.repeat(num_neg)
    neg_dist = F.pairwise_distance(embeddings[src_rep], embeddings[neg_idx])
    neg_loss = F.relu(margin - neg_dist).pow(2).mean()
    return pos_loss + neg_loss

# Training
model     = MusicGNN(input_dim=x.shape[1])
optimizer = torch.optim.Adam(model.parameters(), lr=0.005, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', patience=20, factor=0.5
)

print("\nTraining Advanced GNN...")
print("=" * 50)

best_loss      = float('inf')
best_embedding = None
patience_count = 0
EPOCHS         = 200
EARLY_STOP     = 40

for epoch in range(EPOCHS):
    model.train()
    optimizer.zero_grad()
    embeddings_out = model(x, edge_index)
    loss           = contrastive_loss(embeddings_out, edge_index)
    loss.backward()
    torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
    optimizer.step()
    scheduler.step(loss)

    if loss.item() < best_loss:
        best_loss      = loss.item()
        best_embedding = embeddings_out.detach().numpy().copy()
        patience_count = 0
    else:
        patience_count += 1

    if epoch % 20 == 0:
        print(f"Epoch: {epoch:3d} | Loss: {loss.item():.4f} | Best: {best_loss:.4f}")

    if patience_count >= EARLY_STOP:
        print(f"\nEarly stopping at epoch {epoch}")
        break

print(f"\n✅ Advanced GNN Training Complete!")
print(f"Best loss: {best_loss:.4f}")

# Use best embeddings
embeddings = best_embedding

# Recommendation function
def recommend(song_name, top_n=5):
    if song_name not in song_to_id:
        print("Song not found:", song_name)
        return

    idx        = song_to_id[song_name]
    similarity = cosine_similarity(embeddings[idx].reshape(1, -1), embeddings)[0]
    similarity[idx] = -1

    sorted_indices = similarity.argsort()[::-1]
    selected = []
    used     = set()

    for priority, (col, label) in enumerate([
        ('singer', 'Same Singer'),
        ('genre',  'Same Genre'),
        ('artist', 'Same Artist'),
        ('album',  'Same Album'),
    ]):
        for i in sorted_indices:
            if i != idx and i not in used and df.loc[i, col] == df.loc[idx, col]:
                selected.append((i, label, similarity[i]))
                used.add(i)
                break

    for i in sorted_indices:
        if len(selected) >= top_n:
            break
        if i != idx and i not in used:
            selected.append((i, "Similar Embedding", similarity[i]))
            used.add(i)

    print(f"\n🎵 Recommendations for: {song_name}")
    print("=" * 55)
    for rank, (i, reason, sim) in enumerate(selected, 1):
        print(f"#{rank} {id_to_song[i]}")
        print(f"   Singer : {df.loc[i, 'singer']}")
        print(f"   Genre  : {df.loc[i, 'genre']}")
        print(f"   Album  : {df.loc[i, 'album']}")
        print(f"   Reason : {reason}  |  Similarity: {sim:.3f}")
        print("-" * 55)

recommend("Chamka Chamka")

# Evaluation
def full_evaluation_strict(k=5):
    precisions = []
    recalls    = []

    for song in songs:
        idx = song_to_id[song]
        total_relevant = sum(
            1 for i in range(len(df))
            if i != idx and (
                df.loc[i,'singer'] == df.loc[idx,'singer'] or
                df.loc[i,'album']  == df.loc[idx,'album']
            )
        )

        similarity = cosine_similarity(embeddings[idx].reshape(1,-1), embeddings)[0]
        similarity[idx] = -1
        top_k = similarity.argsort()[::-1][:k]

        hits = sum(
            1 for i in top_k
            if (
                df.loc[i,'singer'] == df.loc[idx,'singer'] or
                df.loc[i,'album']  == df.loc[idx,'album']
            )
        )

        precisions.append(hits / k)
        recalls.append(hits / total_relevant if total_relevant > 0 else 0)

    avg_p = sum(precisions) / len(precisions)
    avg_r = sum(recalls)    / len(recalls)
    f1    = (2 * avg_p * avg_r / (avg_p + avg_r)) if (avg_p + avg_r) > 0 else 0

    print(f"\n📊 Evaluation @ K={k}")
    print("=" * 35)
    print(f"  Precision@{k}  = {avg_p:.3f}")
    print(f"  Recall@{k}     = {avg_r:.3f}")
    print(f"  F1-Score@{k}   = {f1:.3f}")
    print("=" * 35)

full_evaluation_strict(k=5)
full_evaluation_strict(k=10)
````

# 📱 Frontend Integration Guide

## How the Recommendation System Works

### Data Flow:
1. **User plays a song** in the frontend
2. **Frontend sends** song data (title, artist) to ML service API
3. **Model processes** the input and finds the song in dataset
4. **GNN model calculates** similarity scores for all songs
5. **Engine returns** top-k recommendations ranked by:
   - Same Singer (highest priority)
   - Same Genre  
   - Same Artist
   - Same Album
   - Similar Embeddings (fallback)

### API Endpoint:

**POST** `http://localhost:8000/recommend`

### Request Format:

```json
{
    "title": "Chamka Chamka",
    "artist": "Geetha Madhuri",
    "top_k": 5
}
```

### Response Format:

```json
{
    "current_song": {
        "id": 123,
        "title": "Chamka Chamka",
        "singer": "Geetha Madhuri",
        "artist": "Geetha Madhuri",
```

```typescript
async function getRecommendations(songTitle: string, artist: string) {
```

### Similarity Score Scale 📈

- **90-100%**: Extremely similar (same artist/genre/album)
- **70-90%**: Very similar (shares multiple attributes)
- **50-70%**: Moderately similar (complementary style)
- **30-50%**: Somewhat similar (few shared attributes)
- **<30%**: Loosely related (different style)

### Using the Recommendation Hook in Other Components 💻

You can use the `useRecommendations` hook in any component:

```typescript
import { useRecommendations } from '@/hooks/useRecommendations';
import { usePlayer } from '@/context/PlayerContext';

const MyComponent: React.FC = () => {
  const { recommendations, isLoading, error, fetchRecommendations } = useRecommendations();
  const { addToQueue } = usePlayer();

  const handleGetRecommendations = async () => {
    await fetchRecommendations({
      title: "Chamka Chamka",
      artist: "Geetha Madhuri",
      top_k: 8
    });
  };

  return (
    <div>
      <button onClick={handleGetRecommendations} disabled={isLoading}>
        {isLoading ? 'Loading...' : 'Get Recommendations'}
      </button>
      
      {error && <p style={{color: 'red'}}>{error}</p>}
      
      <div>
        {recommendations.map((song) => (
          <div key={song.id}>
            <h3>{song.title}</h3>
            <p>{song.singer}</p>
            <p>Match: {Math.round(song.similarity_score * 100)}%</p>
            <p>{song.recommendation_reason}</p>
            <button onClick={() => addToQueue(song)}>Add to Queue</button>
          </div>
        ))}
      </div>
    </div>
  );
};
```

### Troubleshooting Common Issues 🔧

**Issue: Recommendations not loading**
- [ ] Check ML service is running: `curl http://localhost:8000/`
- [ ] Check browser console (F12) for errors
- [ ] Verify song has title and artist

**Issue: API connection refused**
- [ ] Ensure ML service is running on port 8000
- [ ] Check firewall isn't blocking connections
- [ ] Verify `API_BASE_URL` in hook matches service URL

**Issue: Similarity scores all 0**
- [ ] Check embeddings exist: `ml-services/model/artifacts.pkl`
- [ ] Check model weights: `ml-services/model/gnn_model.pth`
- [ ] May indicate model training issue
